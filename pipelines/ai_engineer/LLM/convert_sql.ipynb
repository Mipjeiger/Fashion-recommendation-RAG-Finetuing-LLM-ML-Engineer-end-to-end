```python
import pandas as pd
import os
from sqlalchemy import inspect, text
from sqlalchemy import create_engine
from dotenv import load_dotenv

load_dotenv()

df = pd.read_parquet("fashion_recommendation")

# ✅ Option 1: Use PostgreSQL (recommended)
DB_USER = os.getenv("DB_USER")
DB_PASSWORD = os.getenv("DB_PASSWORD")
DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT", "5432")
DB_NAME = os.getenv("DB_NAME")

engine = create_engine(
    f"postgresql+psycopg2://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
)
inspector = inspect(engine)
tables = inspector.get_table_names()

with engine.connect() as connection:
    for table in tables:
        result = connection.execute(text(f"SELECT COUNT(*) FROM {table}"))
        row_count = result.fetchone()[0]
        print(f"Table: {table}, Row Count: {row_count}")


print("\n🔍 fashion_recommendation columns:")
columns = inspector.get_columns('fashion_recommendation')
for col in columns:
    print(f"  • {col['name']}: {col['type']}")

# Show table rows into dataframe
with engine.connect() as connection:
    result = connection.execute(text("SELECT * FROM fashion_recommendation"))
    df_db = pd.DataFrame(result.fetchall(), columns=result.keys())
print("\n📊 Sample data from fashion_recommendation:")
df
```
